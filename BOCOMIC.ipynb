```python
import matplotlib.pyplot as plt
import numpy as np




def gaussian(x, mu=0, sigma=1):
    """Normal Gaussian function."""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

x = np.linspace(-4, 4, 500)
dx = x[1]-x[0]
y = np.cumsum(gaussian(x)*dx)
z = np.cumsum(0.5*gaussian(x)*dx+0.5*gaussian(x, mu = 2, sigma = 1)*dx)

# 1. Forzamos a que NO busque las fuentes que te dan error
plt.rcParams['font.family'] = 'sans-serif' 

# 2. Usamos xkcd() pero le decimos que no falle si no hay fuentes
with plt.xkcd(scale=1, length=100, randomness=1):
    plt.figure(figsize=(5,5))

    
    plt.plot([0,0.5,1,0.5,0],[0,1,1,0,0], color = "tab:green", ls='--' )
    plt.plot([0,1],[0,1], color = "tab:gray", ls='--' )
    plt.plot(z/z[-1],y, color = "tab:red")
    plt.scatter((z/z[-1])[300], y[300], color = "black")
    
 
    plt.title("Grafico estilo Comic (Sin error de fuentes)")
    plt.xlabel("$F(X)$")
    plt.ylabel("$F(X|Y=1)$")
    plt.show()
```
