```python
import matplotlib.pyplot as plt
import numpy as np

def gaussian(x, mu=0, sigma=1):
    """Normal Gaussian function."""
    return (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)

x = np.linspace(-4, 4, 500)
dx = x[1]-x[0]


fx0 = gaussian(x,mu=-0, sigma=1)+gaussian(x,mu=-3, sigma=0.5)
fx0 = fx0/np.trapezoid(fx0,x)

fx1 = gaussian(x,mu=-3, sigma=1) + gaussian(x,mu=-4, sigma=1)
fx1 = fx1/np.trapezoid(fx1,x)


#F(X|Y=1)
a = 0.5
fx = a*fx0*dx + (1-a)*fx1*dx
fx1 = fx1*dx
FX = np.cumsum(fx)
FX1 = np.cumsum(fx1)


```

```python
plt.plot(x,fx)
plt.plot(x,fx1)
plt.show()
```

```python
plt.plot(x,FX)
plt.plot(x,FX1)
plt.show()
```

```python
LMR = fx1/fx
plt.plot(x,LMR)
plt.show()
```
